```python
"""
Monitor在每个步骤结束时自动运行，如果 Blue Agent 调用它，则不会产生其他影响。
"""
from pprint import pprint
from CybORG import CybORG
from CybORG.Simulator.Scenarios import EnterpriseScenarioGenerator
from CybORG.Agents import SleepAgent,FiniteStateRedAgent,EnterpriseGreenAgent
from CybORG.Simulator.Actions import Monitor,Analyse


steps=200
sg=EnterpriseScenarioGenerator(
    blue_agent_class=SleepAgent,
    green_agent_class=EnterpriseGreenAgent,
    red_agent_class=SleepAgent,
    steps=steps
)

cyb=CybORG(scenario_generator=sg,seed=1000)

b_name='blue_agent_0'
reset=cyb.reset(agent=b_name)
ini_obs=reset.observation

# pprint(ini_obs)

#session_id agent
action=Monitor(0,b_name)
res = cyb.step(agent=b_name,action=action)

step=1
base_obs=res.observation
# print(f"Step count: {step}")
# pprint(base_obs)

```

```python
new_obs=base_obs

while new_obs == base_obs and step<steps:
    res=cyb.step(agent=b_name,action=action)
    step=step+1
    new_obs=res.observation

print(f"Step count: {step}")
pprint(new_obs)
```

```python
action = Analyse(session=0, agent=b_name, hostname='restricted_zone_a_subnet_user_host_8')

results = cyb.step(agent=b_name, action=action)
print("Step ana1: ", results.observation)
```
