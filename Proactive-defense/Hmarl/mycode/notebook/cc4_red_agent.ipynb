```python
from paramiko.agent import Agent

"""
对于红色代理，有10个可能的操作，该代理从列表中随机选择一个有效的操作。

"""

from pprint import pprint
from CybORG import CybORG
from CybORG.Simulator.Scenarios import EnterpriseScenarioGenerator
from CybORG.Agents import SleepAgent,FiniteStateRedAgent,RandomSelectRedAgent


#初始化环境
steps=100

sg=EnterpriseScenarioGenerator(
    blue_agent_class=SleepAgent,
    green_agent_class=SleepAgent,
    # red_agent_class=RandomSelectRedAgent,
    red_agent_class=FiniteStateRedAgent,
    steps=steps,
)
cyborg=CybORG(scenario_generator=sg,seed=1234)

#记录代理行为
red_agent_0_action=[]

for _ in range(20):
    cyborg.step()
    step_a=cyborg.environment_controller.action
    red_agent_0_action.append(step_a['red_agent_0'])



# pprint(red_agent_0_action)


```
