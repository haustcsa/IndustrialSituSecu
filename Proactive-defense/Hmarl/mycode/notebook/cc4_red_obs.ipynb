```python
from pprint import pprint
from CybORG import CybORG
from CybORG.Simulator.Scenarios import ICSNetworkGenerator
from CybORG.Agents import SleepAgent
from CybORG.Simulator.Actions.AbstractActions import PrivilegeEscalate

steps = 200
sg = ICSNetworkGenerator(blue_agent_class=SleepAgent,
                                green_agent_class=SleepAgent,
                                red_agent_class=SleepAgent,
                                steps=steps)
cyborg = CybORG(scenario_generator=sg, seed=1234)

reset = cyborg.reset(agent='red_agent_5')
first_session_host = list(reset.observation.keys())[1]
initial_obs = reset.observation

print("\nRed Agent 0: Initial Observation \n")
pprint(initial_obs)
"""
通过使用权限升级操作来完成从用户到根级别的访问-权限升级操作
"""


first_action = PrivilegeEscalate(hostname=first_session_host, session=0, agent='red_agent_5')
results = cyborg.step(agent='red_agent_5', action=first_action)
results2 = cyborg.step(agent='red_agent_5', action=first_action)

first_action_obs = results.observation
first_action_obs2 = results2.observation

print("\nRed Agent 3: Observation #1 \n")
pprint(first_action_obs)  #ubuntu
print("\nRed Agent 3: Observation #2 \n")
pprint(first_action_obs2) #root
```
