```python
from CybORG import CybORG
from CybORG.Simulator.Scenarios import ICSNetworkGenerator

"""
CybORG必须通过调用类构造函数手动实例化。
这必须传递给一个ScenarioGenerator类，该类包含场景的详细信息。
"""
sg=ICSNetworkGenerator()
cyborg=CybORG(scenario_generator=sg)

# print(cyborg)
```

```python
from CybORG.Agents.Wrappers import BlueFixedActionWrapper, BlueFlatWrapper
from rich import print
"""
用于枚举每个情节中每个蓝色代理可用的所有操作，同时提供对 CybORG 观察的直接访问
"""

env=BlueFixedActionWrapper(env=cyborg)
obs,_=env.reset()
print(obs.keys())

# print(obs['blue_agent_2'].keys())

# print(env.action_space('blue_agent_0')) #一个整数 操作索引将始终对应于针对特定主机的特定操作
print(env.action_labels('blue_agent_0'))
# print(env.action_mask('blue_agent_0'))
```

```python
"""
要在CybORG环境中执行操作，我们使用env.step（）方法。
此方法采用一个字典，其中键是代理名称，其值是与代理操作空间中的操作相对应的索引。如果指定的动作对于当前事件无效，代理将不做任何事情。
该函数返回下一个观察结果、代理的奖励、每个代理的终止和截断信号以及信息字典。
"""
action={'blue_agent_0':1} #'Restore restricted_zone_a_subnet_user_host_3'
obs,reward,term,trunc,info=env.step(action)
print(reward['blue_agent_0'])
```

```python
"""
提供了一种在代理之间可选地发送8位消息的机制。
这是通过为step函数提供一个代理字典和一个对应的np.数组来实现的，数组中有8个二进制元素。
一个代理发出消息，所以有代理均可以收到
"""
import numpy as np
action={'blue_agent_0':42} #'Restore restricted_zone_a_subnet_user_host_3'
mess={'blue_agent_0' : np.array([0,0,0,0,0,1,0,0])}
obs,reward,term,trunc,info=env.step(action,messages=mess)
print(obs['blue_agent_3']['message'])
```

```python
"""
BlueFlatWrapper来将观察转换为固定大小的向量格式
"""
from CybORG.Agents.Wrappers import BlueFlatWrapper

env = BlueFlatWrapper(env=cyborg)
obs,_=env.reset()

print('Space',len(env.observation_space('blue_agent_0')))
print('Observation:', len(obs['blue_agent_0']))
```
